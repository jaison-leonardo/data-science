```python
import pandas as pd
import dask.dataframe as dd
import time
from google.colab import drive
```

```python
drive.mount('/content/drive')
file_path = "/content/drive/MyDrive/Colab Notebooks/Big data files/flights/flights.csv"
```

Prueba de rendimiento con Pandas

```python
start = time.time()

# cargar dataset
flights_pd = pd.read_csv(file_path)

# limpiar retrasos nulos
flights_pd = flights_pd.dropna(subset=["DEPARTURE_DELAY"])

# promedio de retraso por aerolínea
avg_delay_pd = flights_pd.groupby("AIRLINE")["DEPARTURE_DELAY"].mean()

end = time.time()

print("Resultado Pandas:")
print(avg_delay_pd)

print("\nTiempo Pandas:", round(end - start, 2), "segundos")
```

Prueba de rendimiento con Dask

```python
start = time.time()

# leer dataset con Dask
flights_dd = dd.read_csv(file_path,
                         dtype={'AIR_TIME': 'float64',
                                'ARRIVAL_DELAY': 'float64',
                                'ARRIVAL_TIME': 'float64',
                                'CANCELLATION_REASON': 'object',
                                'DEPARTURE_DELAY': 'float64',
                                'DEPARTURE_TIME': 'float64',
                                'ELAPSED_TIME': 'float64',
                                'SCHEDULED_TIME': 'float64',
                                'TAXI_IN': 'float64',
                                'TAXI_OUT': 'float64',
                                'WHEELS_OFF': 'float64',
                                'WHEELS_ON': 'float64'})

# eliminar nulos
flights_dd = flights_dd.dropna(subset=["DEPARTURE_DELAY"])

# promedio por aerolinea
avg_delay_dd = flights_dd.groupby("AIRLINE")["DEPARTURE_DELAY"].mean()

# ejecutar computo
result = avg_delay_dd.compute()

end = time.time()

print("Resultado Dask:")
print(result)

print("\nTiempo Dask:", round(end - start, 2), "segundos")
```
